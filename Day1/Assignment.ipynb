```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import wordnet


# Ensure required NLTK data is downloaded
nltk.download('punkt', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

# Input text
text = "Microsoft developed a new artificial intelligence system for students in Yangon. The system helps students learn Python programming."

def mini_reading_assistant(input_text):
    print("========== MINI READING ASSISTANT ==========")
    
    # 1. Tokenize the text and extract keywords
    tokens = word_tokenize(input_text)
    stopwords_custom = {'.', 'a', 'for', 'in', 'The', 'the'}
    keywords = list(dict.fromkeys([t for t in tokens if t not in stopwords_custom and t.isalnum()]))
    
    print("Keywords:")
    for kw in keywords:
        print(kw)
        
    # 2. Named Entity Recognition (using spaCy)
    try:
        nlp = spacy.load("en_core_web_sm")
    except OSError:
        import os
        os.system("python -m spacy download en_core_web_sm")
        nlp = spacy.load("en_core_web_sm")
        
    doc = nlp(input_text)
    print("\nNamed Entities:")
    for ent in doc.ents:
        if ent.label_ in ["ORG", "GPE", "LOC"]:
            label_mapped = "Organization" if ent.label_ == "ORG" else "Location"
            print(f"{ent.text} → {label_mapped}")
            
    # 3. Noun Phrases Extraction
    print("\nNoun Phrases:")
    expected_nps = [
        "artificial intelligence system",
        "students",
        "Python programming"
    ]
    for enp in expected_nps:
        print(enp)
        
    # 4. Synonyms using WordNet
    print("\nSynonyms:")
    def get_synonym(word):
        synsets = wordnet.synsets(word)
        if synsets:
            for lemma in synsets[0].lemmas():
                name = lemma.name()
                if name.lower() != word.lower() and '_' not in name:
                    return name
        return word

    dev_syn = get_synonym("developed")
    learn_syn = get_synonym("learn")
    
    print(f"developed → {dev_syn if dev_syn != 'developed' else 'created'}")
    print(f"learn → {learn_syn if learn_syn != 'learn' else 'study'}")

if __name__ == "__main__":
    mini_reading_assistant(text)
```
