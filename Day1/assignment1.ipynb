```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize
from nltk.corpus import stopwords

nltk.download('stopwords')
# Punkt is a pre-trained tokenizer model used by NLTK to help determine where sentences and words are separated.
nltk.download('punkt')
# punkt_tab contains additional tokenizer tables/data used by newer versions of NLTK's Punkt tokenizer.
nltk.download('punkt_tab')

text = "Microsoft developed a new artificial intelligence system for students in Yangon.The system helps students learn Python programming."


tokens = word_tokenize(text)
stopwords_custom = {'.', 'a', 'for', 'in', 'The', 'the','new'}
keywords = list(dict.fromkeys([
    t for t in tokens 
    if t not in stopwords_custom and t.isalnum()]))

print("Keywords:")
for kw in keywords:
    print(kw)

tags = nltk.pos_tag(tokens)

entities = nltk.ne_chunk(tags)
label_mapping = {
    "ORGANIZATION": "Organization",
    "LOCATION": "Location"
}
for subtree in entities:
    if isinstance(subtree, nltk.Tree):
        entity_name = " ".join(word for word, tag in subtree)
        entity_type = label_mapping.get(subtree.label(), subtree.label())
        print(f"{entity_name} -> {entity_type}")

```

```python

grammar = "NP: {<DT>?<JJ.*>*<NN.*>+}"

chunk_parser = nltk.RegexpParser(grammar)
tree = chunk_parser.parse(tags)

for subtree in tree:
    if isinstance(subtree, nltk.Tree) and subtree.label() == "NP":
        np_words = " ".join(word for word, tag in subtree)
        print(np_words)

```

```python
from nltk.corpus import wordnet
nltk.download('wordnet')
nltk.download('omw-1.4')


word = "developed"

synonyms = set()

# Find synsets
for synset in wordnet.synsets(word):

    # Find words in each synset
    for lemma in synset.lemmas():

        synonyms.add(lemma.name())

print("Synonyms:")
print(synonyms)

word = "learn"

synonyms = set()

# Find synsets
for synset in wordnet.synsets(word):

    # Find words in each synset
    for lemma in synset.lemmas():

        synonyms.add(lemma.name())

print("Synonyms:")
print(synonyms)
```
