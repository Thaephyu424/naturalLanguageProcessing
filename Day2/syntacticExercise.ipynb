```python
# ============================================================
# # Topic: Syntactic Parsing using NLTK
# ============================================================

import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')


# ------------------------------------------------------------
# 1. Define the Context-Free Grammar (CFG)
# ------------------------------------------------------------

grammar = nltk.CFG.fromstring("""

    S -> NP VP

    NP -> Det Adj N
    VP -> V NP PP ADVP
    NP -> Det N
    PP -> P NP
    ADVP -> Adv

    Det -> 'the' | 'a'
    Adj -> 'clever'
    N ->  'teacher'|'book' | 'library'
    V -> 'reads'
    P -> 'in'
    Adv -> 'today'

""")


# ------------------------------------------------------------
# 2. Create the Chart Parser
#    ChartParser = a tool that uses CFG rules to build parse trees.
# ------------------------------------------------------------

parser = nltk.ChartParser(grammar)


# ------------------------------------------------------------
# 3. Define the sentence
# ------------------------------------------------------------

text = "the clever teacher reads a book in the library today"
tokens = word_tokenize(text)
print(tokens)
# ------------------------------------------------------------
# 4. Parse the sentence
# ------------------------------------------------------------

for tree in parser.parse(tokens):
    print(tree)

    tree.pretty_print()
```
