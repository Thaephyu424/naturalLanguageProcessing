```python

import nltk
from nltk.tokenize import word_tokenize

sentences = [
    "The student reads a book",
    "The student book reads a"
    ] 

grammar = nltk.CFG.fromstring("""
    S -> NP VP

    NP -> Det N
    VP -> V NP

    Det -> 'The' | 'a'
    N -> 'student' | 'book' 
    V -> 'reads'
""")

parser = nltk.ChartParser(grammar)

for sent in sentences:
    print(f"Sentence:{sent}")

    words = [
        w for w in word_tokenize(sent)
        if w.isalnum()
    ]

    trees = list(parser.parse(words))

    if trees:
        print("Result: Valid Sentence")
        print("parse tree Generated")
        for tree in trees:
            tree.pretty_print()
    else:
        print("Result: Invalid Sentence")
        print("No parse tree found")


  
```
