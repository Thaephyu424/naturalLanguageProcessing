```python
# ============================================================
# Step 1: Import Libraries
# ============================================================

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
```

```python
# ============================================================
# Step 2: Create Dataset
# ============================================================

data = {
    "text": [
        "I love this movie",
        "This movie is amazing",
        "I like this product",
        "This is excellent",
        "I hate this movie",
        "This movie is terrible",
        "I dislike this product",
        "This is bad",
        "Absolute masterpiece, highly recommended!",
        "Complete waste of time and money.",
        "The customer service was exceptionally helpful.",
        "Worst experience I've ever had with a brand.",
        "It exceeded all my expectations.",
        "Broke down after just two days of use.",
        "Such a heartwarming and joyful story.",
        "Extremely disappointing and poorly made."
    ],
    "label": [
        "positive",
        "positive",
        "positive",
        "positive",
        "negative",
        "negative",
        "negative",
        "negative",
        "positive",
        "negative",
        "positive",
        "negative",
        "positive",
        "negative",
        "positive",
        "negative"
    ]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)
```

```python
# ============================================================
# Step 3: Separate X and y
# ============================================================

X = df["text"]
y = df["label"]

print("\nX:")
print(X)

print("\ny:")
print(y)
```

```python
# ============================================================
# Step 4: Split Data
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=25         # Keep the same data split
)

print("\nTraining Data:")
print(X_train)

print("\nTesting Data:")
print(X_test)
```

```python
# ============================================================
# Step 5: Convert Text to TF-IDF
# ============================================================

tfidf = TfidfVectorizer()

X_train_tfidf = tfidf.fit_transform(X_train)    # Learn + convert training data

X_test_tfidf = tfidf.transform(X_test)          # Convert test data

print("\nTraining TF-IDF Shape:")
print(X_train_tfidf.shape)                      # training documents, TF-IDF features

print("\nTesting TF-IDF Shape:")
print(X_test_tfidf.shape)                       # testing documents, TF-IDF features
```

```python
# ============================================================
# Step 6: Create SVM Model
# ============================================================

svm_model = SVC(
    kernel="linear"
)
```

```python
# ============================================================
# Step 7: Train Model
# ============================================================

svm_model.fit(
    X_train_tfidf,
    y_train
)

print("\nModel trained.")
```

```python
# ============================================================
# Step 8: Make Predictions
# ============================================================

y_pred = svm_model.predict(
    X_test_tfidf
)

print("\nActual Labels:")
print(y_test.values)

print("\nPredicted Labels:")
print(y_pred)
```

```python
# ============================================================
# Step 9: Evaluate Model
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\nAccuracy:")
print(accuracy)

```

```python
# ============================================================
# Step 10: Predict New Text
# ============================================================

new_text = [
    "This movie is wonderful"
]

new_vector = tfidf.transform(
    new_text
)

prediction = svm_model.predict(
    new_vector
)

print("\nNew Text:")
print(new_text)

print("\nPrediction:")
print(prediction)
```
